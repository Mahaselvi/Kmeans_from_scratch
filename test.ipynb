### Testing on a synthetic dataset

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from k_means_functions import compute_inertia, kmeans
```

```python
if __name__=='__main__':
    x,y_true=make_blobs(n_samples=500,centers=4,cluster_std=1.0,random_state=42)
    labels,centroids,history=kmeans(x,k=4,seed=1,pp=True)
    print("My inertia : ",round(history[-1],2),f"({len(history)} interactions)")
    km=KMeans(n_clusters=4,init='k-means++',n_init=10,random_state=42).fit(x)
    print("sklearn inertia: ",round(km.inertia_,2))
    fig,ax=plt.subplots(1,2,figsize=(11,4))
    ax[0].scatter(x[:,0],x[:,1],c=labels,s=15)
    ax[0].scatter(centroids[:,0],centroids[:,1],c="red",marker="x",s=200)
    ax[0].set_title("My k-means clusters")
    ax[1].plot(history,marker="o")
    ax[1].set_title("Inertia per iteration")
    ax[1].set_xlabel("iteration")
    plt.tight_layout()
    plt.show()
```
